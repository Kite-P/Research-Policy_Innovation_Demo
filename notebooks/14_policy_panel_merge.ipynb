```python
import pandas as pd

panel = pd.read_parquet('data/processed/research_panel_policy.parquet')
policy = pd.read_parquet('data/processed/province_year_policy_metrics.parquet')
print('enterprise rows/firms/years', len(panel), panel['stock_code'].nunique())
print('enterprise years', panel['year'].min(), panel['year'].max())
print('policy province-year rows', len(policy), policy[['province', 'year']].duplicated().sum())
print('policy_metric_present', panel['policy_metric_present'].value_counts().to_dict())
display(panel.groupby('province')['policy_metric_present'].agg(['size', 'sum']))
policy_cols = ['policy_continuity_tfidf', 'policy_continuity_full_tfidf',
               'policy_continuity_theme']
display(panel[policy_cols].describe())
```

```python
assert len(panel) == 240
assert panel['stock_code'].nunique() == 40
assert panel[['stock_code', 'year']].duplicated().sum() == 0
assert panel['policy_metric_present'].eq(1).all()
assert panel[['province', 'year']].drop_duplicates().shape[0] <= len(panel)
# firm-year 加权分布仅作展示；原始省-年分布见 Notebook 12。
display(panel.groupby(['province', 'year'])['policy_continuity_tfidf'].first().describe())
```
