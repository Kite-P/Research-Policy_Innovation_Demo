# 16 第三章识别设计审计
本 Notebook 只审计样本结构和冻结规格，不输出政策回归系数。

```python
import pandas as pd

panel = pd.read_parquet('data/processed/research_panel_policy.parquet')
specs = pd.read_csv('metadata/chapter3_model_specifications.csv')
print('rows/firms/provinces', len(panel), panel.stock_code.nunique(),
      panel.province.nunique())
print('years', panel.year.min(), panel.year.max())
display(panel.groupby('province').agg(
    firms=('stock_code', 'nunique'), firm_years=('stock_code', 'size')))
audit_cols = ['policy_continuity_tfidf', 'patent_total_ln',
              'invention_ln', 'citation_ln', 'size_ln', 'leverage',
              'roa', 'cash_ratio', 'employee_ln']
display(panel[audit_cols].isna().sum())
display(specs)
```

```python
assert len(panel) == 240 and panel.stock_code.nunique() == 40
assert panel.groupby('stock_code').province.nunique().max() == 1
assert panel.province.nunique() == 7
assert specs.spec_id.tolist() == ['BASE_PRIMARY', 'BASE_SECONDARY_INV', 'BASE_SECONDARY_CIT']
assert specs.primary_inference.eq('webb_wild_cluster_bootstrap').all()
print('No policy regression coefficients are computed in this design audit.')
```
