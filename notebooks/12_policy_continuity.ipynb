# 12 省级政策连续性指标
本 Notebook 只检查政策指标覆盖、分布和轨迹，不加载企业结果变量。

```python
import pandas as pd

metrics = pd.read_parquet('data/processed/province_year_policy_metrics.parquet')
continuity = ['policy_continuity_tfidf', 'policy_continuity_full_tfidf', 'policy_continuity_theme']
print('coverage by year')
display(metrics.groupby('year')[continuity].count())
print('distribution')
display(metrics[continuity].describe(percentiles=[.01, .05, .25, .5, .75, .95, .99]))
```

```python
print('province trajectories')
display(metrics.pivot(index='province', columns='year', values='policy_continuity_tfidf'))
print('quality flags')
display(metrics['metric_quality_flag'].value_counts())
assert metrics[['province', 'year']].duplicated().sum() == 0
assert metrics.loc[metrics.year == 2019, continuity].isna().all().all()
assert metrics.loc[metrics.year >= 2020, continuity].notna().all().all()
assert metrics[continuity].dropna().apply(lambda col: col.between(0, 1).all()).all()
```
