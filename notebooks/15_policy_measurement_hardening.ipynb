# 15 政策来源与连续性测量加固

```python
import pandas as pd

manifest = pd.read_csv('metadata/policy_source_manifest.csv')
metrics = pd.read_parquet('data/processed/province_year_policy_metrics.parquet')
revision = pd.read_csv('results/policy_continuity/measurement_revision.csv')
display(manifest.groupby(['source_tier', 'source_tier_label']).size())
measurement_cols = ['policy_continuity_tfidf',
                    'policy_continuity_tfidf_expanding',
                    'source_tier_max', 'source_tier_changed',
                    'both_direct_official']
display(metrics[measurement_cols].describe())
display(revision[['absolute_delta']].describe())
assert len(manifest) == 49 and manifest['source_tier_verified'].eq(1).all()
assert metrics.loc[metrics.year >= 2020, 'policy_continuity_tfidf_expanding'].notna().all()
assert revision['absolute_delta'].fillna(0).eq(0).all()
```

```python
pair_cols = ['pair_mean_log_industry_chars',
             'abs_log_industry_length_change',
             'pair_mean_log_full_chars', 'abs_log_full_length_change']
display(metrics[pair_cols].describe())
assert metrics.loc[metrics.year == 2019, pair_cols].isna().all().all()
assert metrics.loc[metrics.year >= 2020, pair_cols].notna().all().all()
```
